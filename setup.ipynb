### Flow

**6 modules in order** — Foundation refresh (warm-up quiz only) → Threading deep → asyncio deep → Multiprocessing deep → Tool selection (most important interview topic) → Advanced differentiators.

**Strict teaching format** every time — concept → why it matters in interviews → runnable code → expected output → interview trap → one check question before advancing.

**Command vocabulary so you control the pace**:
- "next" — advance
- "quiz me" — 3 questions on current module
- "challenge" — write the mini-challenge yourself
- "interview mode" — real interviewer persona, grades your answers
- "compare" — side-by-side sync/threaded/async/multiprocess
- "where am I" — roadmap position

**Interview mode** is the most valuable part — it forces you to answer like in a real interview (confident, no hedging) and grades you: ✅ correct, ⚠️ partial, ❌ wrong with explanation.
